#### Grouped:
- which ,eans based on the particular non-numeric value, we will going to analyse the numerical column in a particular DataSet.

```python
## We will find the mean of value based on the product column which has 3 Products

grouped_mean = df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby('Category')['Product'].sum()
print(grouped_sum)
```

```python
multi_columns = df.groupby(['Product','Region'])['Sales'].sum()
print(multi_columns)
```

```python
## Aggregate Functions

grouped_age = df.groupby('Product')['Sales'].agg(['sum','count','mean'])
print(grouped_age)
```

```python
grouped = df.groupby(['Product','Region'])['Value'].agg(['sum','count','mean'])
print(grouped)
```

```python
## Merging and Joining DataFrames

## create Sample DataFrames

df1 = pd.DataFrame({'key':['A','B','C'],'VALUES1':[1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'],'VALUES2':[4,5,6]})

```

```python
## Merge DataFrame on the ketColumn

pd.merge(df1,df2,on='key',how='inner')
```

```python
pd.merge(df1,df2,on='key',how='outer')
```

```python
pd.merge(df1,df2,on='key',how="left")
```

```python
pd.merge(df1,df2,on='key',how='right')
```
