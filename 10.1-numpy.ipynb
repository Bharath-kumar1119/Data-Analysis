### Attributes of NUMPY

```python
arr = np.array(([1,2,3],[4,5.2,6]))
print("Array: \n", arr)
print("No.of dimension:", arr.ndim)
print("number of elements:",arr.size)
print("Datatype:",arr.dtype)
print("Item size: ",arr.itemsize)
```

### Numpy Vectorized Operation

```python
Arr = np.array([1,2,3,4,5,6])
Arr2 = np.array([10,20,30,40,50,60])

## Element wise Addition

print("Addition is: ",Arr+Arr2)

## Element Wise Subtraction

print("Subtraction: ",Arr-Arr2)
print("Subtraction: ",Arr2*Arr)

print(Arr.sum())
print(Arr2.sum())
print("Sum of two arrays are:",Arr.sum()+Arr2.sum())
```

### Universal Functions

```python
arr = np.array([25,64,36])
print(np.sqrt(arr))
print(np.exp(arr))
print(np.sin(arr))
```

### Normalization
- which has mean 0 and standard deviation 1

```python
data = np.array([70,80,90,100])

mean = np.mean(data)
print(mean)

std = np.std(data)
print(std)

normal_devoiation = (data - mean) / std
print(normal_devoiation)
```

```python
data = np.array([10,20,30])
print(data-5)
print(data+5)
print(data/5)
print(data//5)
print(data%5)
```
