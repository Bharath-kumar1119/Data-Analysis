### Series
- A panda series is a 1D array-like object that can hold any data type.
- It is similar to column in a table.

```python
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series: \n",series)
```

```python
## create a Series from dictionary
data1 = {'a':1,'b':2}
series_dict = pd.Series(data1)
print(series_dict)
```

```python
data2 = {1:"10",2:2.2,3:30,4:"bharath"}
series_data2 = pd.Series(data2)
print(series_data2)
```

```python
data = [16,26,36]
indexes = ['a','b','c']
pd.Series(data,index=indexes)
```

### DataFrame
- 2 Dimensional
- multiple rows and columns

```python
## DataFrame from the dictionary

data = {
    'studentName' : ['bharath','Surya','Sita'],
    'Age' : [18,17,18],
    'City' : ['Vadodara','narsipatnam','Srikakulam']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
Animal = {
    'AnimalName':['Dog','Cat','BUffalo','Cow'],
    'Expansion' : [10,7,15,25],
}

df1 = pd.DataFrame(Animal)
print(df1)
df1.size
```
